**## Step 1: Import Libraries**



```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

**## Step 2: Load Dataset**

```python
df = pd.read_csv('/content/Mall_customers.csv')
```

```python
df.head()
```

```python
df.info()
```

```python
df.describe()
```

```python
df.isnull().sum()
```

```python
print("Duplicate rows:", df.duplicated().sum())
df.drop_duplicates(inplace=True)
print("Shape after cleaning:", df.shape)
```

**The dataset has no missing values and no duplicate rows. It is clean and ready for analysis.**

```python
sns.countplot(x='Gender', data=df)

plt.show()
```

**Observation**: The dataset contains slightly more female customers than male customers.

```python
sns.histplot(df['Age'],bins=20)
plt.show()
```

**Observation**: Most customers belong to the age group between 20 and 40 years.

```python
sns.boxplot(data=df[['Age', 'Annual Income (k$)', 'Spending Score (1-100)']])
plt.title("Boxplot - Outlier Detection")
plt.show()
```

**Observation**: Age and Spending Score look clean with no major outliers. A few high values in Annual Income are normal and valid.


```python
sns.scatterplot(x='Annual Income (k$)', y='Spending Score (1-100)', data=df)
plt.show()
```

**Observation**: Customers with similar annual income often show very different spending habits. Some customers with moderate income spend a lot, while some high-income customers spend less.

```python
numeric_df = df.select_dtypes(include='number')

plt.figure(figsize=(8,5))

sns.heatmap(numeric_df.corr(), annot=True)

plt.show()
```

**Observation**: The heatmap shows that a customer’s income does not strongly affect their spending habits. Some customers with higher income spend less, while others with moderate income spend more.
